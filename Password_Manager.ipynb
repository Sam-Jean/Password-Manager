```python
import sys

usernames = {}
while True:
  print("press 1 to create new username")
  print("press 2 to view a username")
  print("press 3 to quit")
  userinput = input("press 1,2 or 3")


  if userinput == "3":
    sys.exit()

  if userinput == "1":
    username = input("please enter a username")
    url = input("please enter a url")
    password = input("please enter a password")
    userurll = [url, username]
    userurl = tuple(userurll)

    usernames[userurl] = password
    print(usernames)

  if userinput == "2":
    searchkey = input("please enter the url or username you want to search with")
    keys = usernames.keys()
    for key in keys:
      if searchkey in key[0] or searchkey in key[1]:
          print(key, usernames[key])



```
